```python
jumlah_buku = int(input("Masukkan jumlah buku yang dibeli:  "))
harga_per_buku = float(input("Masukkan harga per buku:  "))
uang_budi = float(input("Masukkan uang yang dibayarkan:  "))
jumlah_orang = int(input("Masukkan jumlah orang untuk patungan:  "))

total_belanja = jumlah_buku * harga_per_buku
diskon = 0.10 * total_belanja
jumlah_bayar = total_belanja - diskon
uang_kembalian = uang_budi - jumlah_bayar
iuran_per_orang = int(jumlah_bayar // jumlah_orang)
sisa_patungan = jumlah_bayar % jumlah_orang
uang_cukup = uang_budi >= jumlah_bayar

print(f"Total Belanja: Rp {total_belanja:,.0f}")
print(f"Besar Diskon (10%): Rp {diskon:,.0f}")
print(f"Jumlah yang Harus Dibayar: Rp {jumlah_bayar:,.0f}")
print(f"Uang Kembalian Budi: Rp {uang_kembalian:,.0f}")
print(f"Iuran per Orang ({jumlah_orang} orang): Rp {iuran_per_orang:,.0f}")
print(f"Sisa / Selisih Patungan: Rp {sisa_patungan:,.0f}")
print(f"Apakah Uang Budi Cukup?: {uang_cukup}")
```

```python
jumlah_panel = int(input("Masukkan jumlah panel surya: "))
daya_wp = float(input("Masukkan daya per panel (Wp): "))
jam_penyinaran = float(input("Masukkan rata-rata penyinaran (jam/hari): "))
efisiensi = float(input("Masukkan efisiensi sistem (misal 0.8): "))
kebutuhan_listrik = float(input("Masukkan kebutuhan listrik rumah (kWh/hari): "))
tarif_listrik = float(input("Masukkan tarif listrik (Rp/kWh): "))

energi_harian = round((jumlah_panel * daya_wp * jam_penyinaran * efisiensi) / 1000, 1)
energi_bulanan = round(energi_harian * 30, 1)
nilai_listrik = round(energi_bulanan * tarif_listrik, 1)
kekurangan = round(kebutuhan_listrik - energi_harian, 1)
mencukupi = energi_harian >= kebutuhan_listrik

print(f"Energi harian (kWh) : {energi_harian}")
print(f"Energi bulanan (kWh): {energi_bulanan}")
print(f"Nilai listrik (Rp)  : {nilai_listrik}")
print(f"Kekurangan (kWh)    : {kekurangan}")
print(f"Mencukupi?          : {mencukupi}")
```

```python
daya_generator = float(input("Masukkan daya generator (kW): "))
jam_operasi = float(input("Masukkan jam operasi per hari: "))
konsumsi_bbm = float(input("Masukkan konsumsi bahan bakar (liter/kWh): "))
harga_solar = float(input("Masukkan harga solar per liter (Rp): "))
kapasitas_tangki = float(input("Masukkan kapasitas tangki (liter): "))
arus = float(input("Masukkan arus pada kabel (A): "))
hambatan = float(input("Masukkan hambatan kabel (Ohm): "))

energi = daya_generator * jam_operasi
solar_harian = energi * konsumsi_bbm
biaya = solar_harian * harga_solar
tangki_cukup = kapasitas_tangki // solar_harian
sisa_solar = kapasitas_tangki % solar_harian
rugi_kabel = (arus ** 2) * hambatan
rugi_tinggi = (rugi_kabel > 150) and (arus >= 20)

print(f"\nEnergi (kWh)      : {energi}")
print(f"Solar (liter)     : {solar_harian}")
print(f"Biaya (Rp)        : {biaya}")
print(f"Tangki cukup      : {float(tangki_cukup)} hari")
print(f"Sisa solar (L)    : {float(sisa_solar)}")
print(f"Rugi kabel (W)    : {rugi_kabel}")
print(f"Rugi tinggi?      : {rugi_tinggi}")
```
