```python
tegangan_harian = [217.5, 219.0, 220.3, 218.7, 221.1, 216.9]

# (a) Cetak pembacaan pertama dan terakhir memakai indeks
print("Pembacaan pertama:", tegangan_harian[0])
print("Pembacaan terakhir:", tegangan_harian[-1])

# (b) Tambahkan satu pembacaan baru 222.0 dengan append()
tegangan_harian.append(222.0)

# (c) Cetak jumlah total data dengan len()
print("Jumlah total data:", len(tegangan_harian))
```

```python
# 1. Membuat dictionary spesifikasi motor listrik
motor_listrik = {
    "nama": "EM-Induction-X1",
    "daya_kW": 15.0,
    "tegangan_V": 380,
    "rpm": 1450
}

# 2. Mencetak salah satu nilai lewat key-nya (contoh: daya_kW)
print("Daya motor listrik:", motor_listrik["daya_kW"], "kW")

# 3. Mengubah nilai rpm menggunakan operator '='
motor_listrik["rpm"] = 3000

# 4. Mencetak ulang seluruh dictionary
print("\nSpesifikasi motor listrik setelah diperbarui:")
print(motor_listrik)
```

```python
pembangkit = [
    {"nama": "Unit-1", "kapasitas_MW": 100, "status": "operasi"},
    {"nama": "Unit-2", "kapasitas_MW": 150, "status": "operasi"},
    {"nama": "Unit-3", "kapasitas_MW": 80,  "status": "pemeliharaan"},
]

# 1. Cetak nama unit kedua (indeks 1)
print("Nama unit kedua:", pembangkit[1]["nama"])

# 2. Ubah status salah satu unit (misal Unit-3)
pembangkit[2]["status"] = "operasi"

# 3. Tambahkan satu unit baru menggunakan append()
pembangkit.append({"nama": "Unit-4", "kapasitas_MW": 200, "status": "pemeliharaan"})

# 4. Cetak seluruh data
print("\nData seluruh pembangkit:")
for unit in pembangkit:
    print(unit)
```
